```python
import pandas as pd
from google.colab import files
import numpy as np
import matplotlib.pyplot as plt
import joblib
from sklearn.preprocessing import MinMaxScaler
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score
from sklearn.pipeline import Pipeline
```

```python
uploaded = files.upload()
arquivo = next(iter(uploaded))
df = pd.read_csv(arquivo)
print(df['label'].value_counts())
print(df.shape)
```

```python
X,y = df.drop(['label','codigo'], axis=1), df['label']
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.3)
```

```python
pipe_tree = Pipeline([
    ('minmax',MinMaxScaler()),
    ('dtc', DecisionTreeClassifier(max_depth=1)),
    ])
pipe_rf = Pipeline([
    ('minmax',MinMaxScaler()),
    ('rfc', RandomForestClassifier( n_estimators=50,random_state=42, max_depth=1))
])
pipe_knn = Pipeline([
    ('minmax', MinMaxScaler()),
    ('knn', KNeighborsClassifier(n_neighbors=5))
])
```

```python
pipe_tree.fit(X_train,y_train)
pipe_tree.predict(X_test)
```

```python
pipe_rf.fit(X_train,y_train)
pipe_rf.predict(X_test)
```

```python
pipe_knn.fit(X_train,y_train)
pipe_knn.predict(X_test)
```

```python
print(classification_report(y_test,pipe_tree.predict(X_test),target_names=['Formado','Não Formado']))
```

```python
print(classification_report(y_test,pipe_rf.predict(X_test),target_names=['Formado','Não Formado']))
```

```python
print(classification_report(y_test,pipe_knn.predict(X_test),target_names=['Formado','Não Formado']))
```

```python
y_pred_treino_tree = pipe_tree.predict(X_train)
print("Acurácia no Treino:", accuracy_score(y_train, y_pred_treino_tree))
print("Acurácia no Teste: ", accuracy_score(y_test, pipe_tree.predict(X_test)))
y_pred_treino_rf = pipe_rf.predict(X_train)
print("Acurácia no Treino:", accuracy_score(y_train, y_pred_treino_rf))
print("Acurácia no Teste: ", accuracy_score(y_test, pipe_rf.predict(X_test)))
y_pred_treino_knn = pipe_knn.predict(X_train)
print("Acurácia no Treino:", accuracy_score(y_train, y_pred_treino_knn))
print("Acurácia no Teste: ", accuracy_score(y_test, pipe_knn.predict(X_test)))
```

```python
scores_tree = cross_val_score(pipe_tree, X, y, cv=5, scoring='recall')
scores_rf = cross_val_score(pipe_rf, X, y, cv=5, scoring='recall')
scores_knn = cross_val_score(pipe_knn, X, y, cv=5, scoring='recall')
```

```python
valores_tree = cross_val_score(pipe_tree,X,y,scoring = 'recall',cv=5)
valores_rf = cross_val_score(pipe_rf,X,y,scoring = 'recall',cv=5)
valores_knn = cross_val_score(pipe_knn,X,y,scoring = 'recall',cv=5)
```

```python
media_tree = np.mean(valores_tree)
dpa_tree = np.std(valores_tree)
print(f'Média:',media_tree)
print(f'Desvio Padrão:',dpa_tree)
media_rf = np.mean(valores_rf)
dpa_rf = np.std(valores_rf)
print(f'Média:',media_rf)
print(f'Desvio Padrão:',dpa_rf)
media_knn = np.mean(valores_knn)
dpa_knn = np.std(valores_knn)
print(f'Média:',media_knn)
print(f'Desvio Padrão:',dpa_knn)
```
